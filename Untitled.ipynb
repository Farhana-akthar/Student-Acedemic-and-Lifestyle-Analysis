```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Optional: make graphs appear inside notebook
%matplotlib inline
```

```python
# Create dataset with realistic names
data = pd.DataFrame({
    "Student_ID": ["Fahad Khan","Ananya Sharma","Rohan Mehta","Sanya Iyer","Aditya Verma",
                   "Isha Patel","Vikram Singh","Meera Reddy","Karan Gupta","Sneha Joshi"],
    "Age": [18, 19, 20, 21, 18, 19, 20, 21, 19, 18],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female"],
    "Study_Hours": [5, 6, 3, 4, 7, 2, 5, 4, 6, 3],
    "Sleep_Hours": [7, 8, 6, 7, 8, 6, 7, 7, 8, 6],
    "Attendance": [85, 92, 70, 80, 95, 65, 88, 75, 90, 72],
    "GPA": [3.2, 3.8, 2.5, 3.0, 3.9, 2.2, 3.5, 2.9, 3.6, 2.7],
    "Extracurricular": ["Yes","No","Yes","No","Yes","No","Yes","No","Yes","No"],
    "Internet_Usage": [3, 2, 5, 4, 3, 6, 2, 4, 3, 5],
    "Social_Media_Hours": [2,1,4,3,1,5,2,3,1,4]
})

# Show the dataset
data
```

```python
# Basic info about the dataset
data.info()
data.describe()
data.isnull().sum()
```

```python
sns.scatterplot(data=data, x="Study_Hours", y="GPA", hue="Gender", s=100)
plt.title("GPA vs Study Hours")
plt.show()
```

```python
sns.scatterplot(data=data, x="Sleep_Hours", y="GPA", hue="Extracurricular", s=100)
plt.title("GPA vs Sleep Hours")
plt.show()
```

```python
sns.scatterplot(data=data, x="Attendance", y="GPA", hue="Gender", s=100)
plt.title("GPA vs Attendance")
plt.show()
```

```python
sns.barplot(data=data, x="Student_ID", y="Social_Media_Hours")
plt.xticks(rotation=45)
plt.title("Social Media Hours per Student")
plt.show()
```

```python
for index, row in data.iterrows():
    print(f"Analysis for {row['Student_ID']}:")
    
    if row["GPA"] >= 3.5:
        print("- GPA is excellent ✅")
    elif row["GPA"] >= 3.0:
        print("- GPA is good 🙂")
    else:
        print("- GPA needs improvement ⚠️")
    
    if row["Sleep_Hours"] >= 7:
        print("- Good sleep habits 😴")
    else:
        print("- Needs more sleep ⏰")
        
    if row["Extracurricular"] == "Yes":
        print("- Active in extracurricular activities 🎨⚽")
    else:
        print("- Not active in extracurricular activities ❌")
    
    if row["Internet_Usage"] > 4:
        print("- High internet usage 💻")
    else:
        print("- Moderate internet usage 👍")
    
    print("\n")
```

```python
data.to_csv("students_analysis_output.csv", index=False)
```
